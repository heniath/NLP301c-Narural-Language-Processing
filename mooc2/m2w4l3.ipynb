```python
word = 'dearz'
```

```python
splits_a = []
for i in range(len(word) + 1):
    splits_a.append([word[:i], word[i:]])
for i in splits_a:
    print(i)     
```

```python
splits_b = [([word[:i], word[i:]] for i in range(len(word) + 1))]
for i in range(len(splits_b)):
    print(i)
```

```python
splits = splits_a
deletes = []
print(f'word: {word}')
for L, R in splits:
    if R:
        print(f'{L + R[1:]}. Deleted {R[0]}')
```

```python
splits = splits_a
deletes = [L + R[1:] for L, R in splits]
print(deletes)
for i in deletes:
    print(i)
```

```python
vocab = ['dean','deer','dear','fries','and','coke']
edit = list(deletes)

candidates = set(vocab).intersection(set(edit))
print(candidates)
```
