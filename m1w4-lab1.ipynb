```python
angle = 100 * (np.pi / 180)

Ro = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle), np.cos(angle)]
])

x2 = np.array([[2, 2]])
y2 = np.dot(Ro, x2.T)

print(Ro)
print(y2)

print('\n')

print(np.linalg.norm(x2))
print(np.linalg.norm(y2))
print(np.linalg.norm(Ro))
```

```python
plot_vectors([x2, y2], axes=[4, 4], fname='transform2.svg')
```
