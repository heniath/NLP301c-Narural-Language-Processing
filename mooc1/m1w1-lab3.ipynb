```python
data = pd.read_csv('./data/logistic_features.csv')
print(data.head())
print(data.tail())
```

```python
X = data[['bias', 'positive', 'negative']].to_numpy()
Y = data['sentiment'].to_numpy()

print(X.shape)
print(X)
```

```python
theta = [6.03518871e-08, 5.38184972e-04, -5.58300168e-04]
```

```python
fig, ax = plt.subplots(figsize=(8, 8))

colors = ['red', 'green']

ax.scatter(X[:, 1], X[:, 2], c=[colors[int(k)] for k in Y], s=0.1)
plt.xlabel('Positive')
plt.ylabel('Negative')
```

```python
def neg(theta, pos):
    return (-theta[0] - pos * theta[1]) / theta[2]

def direction(theta, pos):
    return pos * theta[2] / theta[1]
```

```python
fix, ax = plt.subplots(figsize=(8, 8))

colors = ['red', 'green']

ax.scatter(X[:, 1], X[:, 2], c=[colors[int(k)] for k in Y], s=0.5)
plt.xlabel('Positive')
plt.ylabel('Negative')

max_pos = np.max(X[:, 1])
offset = 5000

ax.plot([0, max_pos], [neg(theta, 0), neg(theta, max_pos)], c='gray')
ax.arrow(offset, neg(theta, offset), offset, direction(theta, offset), head_width=500, head_length=500, fc='g', ec='g')
ax.arrow(offset, neg(theta, offset), -offset, -direction(theta, offset), head_width=500, head_length=500, fc='r', ec='r')

plt.show()
```
